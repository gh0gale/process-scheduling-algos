```python
def round_robin_scheduling(processes, time_quantum):
    n = len(processes)
    processes.sort(key=lambda x: x[1])  # Sort by arrival time
    remaining_time = {p[0]: p[2] for p in processes}
    current_time = 0
    waiting_time = {p[0]: 0 for p in processes}
    last_execution = {p[0]: p[1] for p in processes}
    
    turnaround_time = {p[0]: 0 for p in processes}
    queue = []

    while processes or queue:
        # Add processes that have arrived to the queue
        while processes and processes[0][1] <= current_time:
            queue.append(processes.pop(0))

        if queue:
            # Process to execute next
            pid, arrival, burst = queue.pop(0)
            if last_execution[pid] > current_time:
                waiting_time[pid] += last_execution[pid] - current_time
                current_time = last_execution[pid]

            # Execute for at most the time quantum or until completion
            execution_time = min(time_quantum, remaining_time[pid])
            remaining_time[pid] -= execution_time
            current_time += execution_time
            last_execution[pid] = current_time

            if remaining_time[pid] == 0:
                # Process completes
                turnaround_time[pid] = current_time - arrival
                print(f"Process {pid}: Completion={current_time}, Turnaround={turnaround_time[pid]}, Waiting={waiting_time[pid]}")
            else:
                # Add back to the queue
                queue.append((pid, arrival, burst))

        else:
            # No process is ready to execute
            current_time += 1

    avg_waiting_time = sum(waiting_time.values()) / n
    avg_turnaround_time = sum(turnaround_time.values()) / n
    print(f"Average Waiting Time: {avg_waiting_time:.2f}")
    print(f"Average Turnaround Time: {avg_turnaround_time:.2f}")

# Example
processes = [("P1", 0, 5), ("P2", 1, 3), ("P3", 2, 8), ("P4", 3, 6)]
time_quantum = 2
round_robin_scheduling(processes, time_quantum)
```
