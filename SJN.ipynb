```python
def sjn_scheduling(processes):
    processes.sort(key=lambda x: x[1])  # Sort by arrival time
    n = len(processes)
    completed = []
    current_time = 0
    waiting_time = []
    turnaround_time = []
   

    while processes:
        ready_queue = [p for p in processes if p[1] <= current_time]
        if ready_queue:
            next_process = min(ready_queue, key=lambda x: x[2])  # Pick process with shortest burst time
            processes.remove(next_process)
            pid, arrival, burst = next_process
            start_time = current_time
            completion_time = start_time + burst
            turnaround = completion_time - arrival
            wait = turnaround - burst
            waiting_time.append(wait)
            turnaround_time.append(turnaround)
            completed.append(pid)
            current_time += burst
            print(f"Process {pid}: Start={start_time}, Completion={completion_time}, Waiting={wait}, Turnaround={turnaround}")
        else:
            current_time += 1

    avg_waiting_time = sum(waiting_time) / n
    avg_turnaround_time = sum(turnaround_time) / n
    print(f"Average Waiting Time: {avg_waiting_time:.2f}")
    print(f"Average Turnaround Time: {avg_turnaround_time:.2f}")

# Example
processes = [("P1", 0, 7), ("P2", 2, 4), ("P3", 4, 1), ("P4", 5, 4)]
sjn_scheduling(processes)
```
