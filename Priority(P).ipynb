```python
def preemptive_priority_scheduling(processes):
    n = len(processes)
    processes.sort(key=lambda x: x[1])  # Sort by arrival time
    remaining_time = {p[0]: p[2] for p in processes}
    current_time = 0
    completed = 0
    waiting_time = {p[0]: 0 for p in processes}
    last_execution = {p[0]: 0 for p in processes}

    while completed < n:
        ready_queue = [p for p in processes if p[1] <= current_time and remaining_time[p[0]] > 0]
        if ready_queue:
            next_process = min(ready_queue, key=lambda x: x[3])  # Pick process with highest priority (lowest value)
            pid, arrival, burst, priority = next_process

            # Execute the process for 1 unit of time
            if last_execution[pid] > 0:
                waiting_time[pid] += current_time - last_execution[pid]
            remaining_time[pid] -= 1
            last_execution[pid] = current_time + 1

            if remaining_time[pid] == 0:
                completed += 1
                turnaround_time = current_time + 1 - arrival
                print(f"Process {pid}: Completion={current_time + 1}, Turnaround={turnaround_time}, Waiting={waiting_time[pid]}")

            current_time += 1
        else:
            current_time += 1  # Idle time

    avg_waiting_time = sum(waiting_time.values()) / n
    avg_turnaround_time = avg_waiting_time + sum(p[2] for p in processes) / n
    print(f"Average Waiting Time: {avg_waiting_time:.2f}")
    print(f"Average Turnaround Time: {avg_turnaround_time:.2f}")

# Example
processes = [("P1", 0, 5, 2), ("P2", 1, 3, 1), ("P3", 2, 8, 4), ("P4", 3, 6, 3)]
preemptive_priority_scheduling(processes)
```
